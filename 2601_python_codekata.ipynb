```python
# Q1. 두 수의 차
# 정수 num1과 num2가 주어질 때, num1에서 num2를 뺀 값을 return하도록 soltuion 함수를 완성해주세요.
# 제한사항
# #-50000 ≤ num1 ≤ 50000
# -50000 ≤ num2 ≤ 50000


def solution(num1, num2):
    answer = num1 - num2
    return answer
```

# 파이썬의 기초 연산자인 -를 활용함. 
# 정수 범위 제한이 있지만 파이썬은 자동으로 처리해주어 별도 예외처리는 하지 않음.

```python
# Q2. 두 수의 곱
# 정수 num1, num2가 매개변수 주어집니다. num1과 num2를 곱한 값을 return 하도록 solution 함수를 완성해주세요.
# 제한사항
# 0 ≤ num1 ≤ 100
# 0 ≤ num2 ≤ 100


def solution(num1, num2):
    answer = num1 * num2
    return answer
```

# 파이썬의 기초 연산자인 *를 활용함. 
# 정수 범위 제한이 있지만 파이썬은 자동으로 처리해주어 별도 예외처리는 하지 않음.

```python
# Q3. 몫 구하기
# 정수 num1, num2가 매개변수로 주어질 때, num1을 num2로 나눈 몫을 return 하도록 solution 함수를 완성해주세요.
# 제한사항
# 0 < num1 ≤ 100
# 0 < num2 ≤ 100


def solution(num1, num2):
    answer = num1 // num2
    return answer
```

```python
# Q4. 나이 출력
# 머쓱이는 선생님이 몇 년도에 태어났는지 궁금해졌습니다.
# 2022년 기준 선생님의 나이 age가 주어질 때, 선생님의 출생 연도를 return 하는 solution 함수를 완성해주세요
# 제한사항
# 0 < age ≤ 120
# 나이는 태어난 연도에 1살이며 매년 1월 1일마다 1살씩 증가합니다.


def solution(age):
    answer = 2022 - age + 1
    return answer
```

# 나이 구하는 공식 = 현재 년도 - 나이 + 1 → 태어날 때부터 1살로 시작하니까

```python
# Q5. 숫자 비교하기
# 정수 num1과 num2가 매개변수로 주어집니다. 두 수가 같으면 1 다르면 -1을 retrun하도록 solution 함수를 완성해주세요.
# 제한사항
# 0 ≤ num1 ≤ 10,000
# 0 ≤ num2 ≤ 10,000


def solution(num1, num2):
    if num1 == num2 :
        answer = 1
    else :
        answer = -1
    return answer
```
